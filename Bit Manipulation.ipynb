```python
20 = 16 + 4     = 0001 0100 (Even)
21 = 16 + 4 + 1 = 0001 0101 (Odd)
                & 0000 0001
                  ---------
                  0000 0001 (Non-Zero => ODD)
```

```python
def isOdd(n):
    return ((n & 1) == 1)
print(isOdd(20))
print(isOdd(21))
```

```python
def isEven(n):
    return not isOdd(n)
print(isEven(20))
print(isEven(21))
```

```python
Finding the i-th bit of the number n
                   7654 3210
n = 20 => 16 + 4 = 0001 0100
i = 2            & 0000 0100 => mask => 1 << 2 = 1 << i
                   ---------
                   0000 0100 != 0 => at the i-th place the bit is 1
                                7654 3210
n = 55 => 32 + 16 + 4 + 2 + 1 = 0011 0111
i = 5                         & 0010 0000 => mask => 1 << 5 = 1 << i
                                ---------
                                0010 0000 != 0 => at the i-th place the bit is 1
```

```python
def getBit(n, i):
    mask = 1 << i
    bit = 1 if (n & mask) else 0
    return bit
print(getBit(20, 2))
print(getBit(55, 5))
print(getBit(55, 3))
```

```python
Set the i-th bit of the number n
set/preset means making the bit 1
reset/clear means making the bit 0
                   7654 3210
n = 20 => 16 + 4 = 0001 0100
i = 3            | 0000 1000 => mask => 1 << 3 = 1 << i
                   ---------
                   0001 1100 => 16 + 8 + 4 = 28
```

```python
def setBit(n, i):
    mask = 1 << i
    return n | mask
print (setBit(20, 3))
print (setBit(20, 5))
```

```python
Clear the i-th bit of the number n
                   7654 3210
n = 20 => 16 + 4 = 0001 0100
i = 2            & 1111 1011 => mask => ~(1 << 2) = ~(0000 0100) = 1111 1011 = ~(1 << i)
                   ---------
                   0001 0000 => 16
```

```python
def clearBit(n, i):
    mask = ~(1 << i)
    return n & mask
print(clearBit(20, 2))
print(clearBit(55, 5))
print(clearBit(55, 3))
```

```python
clearing last i bits from the right end of the number n
clearLastIBits(1011 1111, 4) => 1011 0000
                   7654 3210
n = 20 => 16 + 4 = 0001 0100
i = 4            & 1111 0000 => mask => (-1) << 4 = (1111 1111) << 4 = 1111 0000 = (-1) << i
                   ---------
                   0001 0000 => 16
n = 55 => 32 + 16 + 4 + 2 + 1 = 0011 0111
Conclusion: -1 = ~0 = 1111 1111
```

```python
def clearLastIBits(n, i):
    mask = -1 << i
    return n & mask
print(clearLastIBits(20, 4))
print(clearLastIBits(55, 4))
```

```python
Clear all bits from the number n within the range i (left limit) to j (right limit)
n = 63, i = 2, j = 4
                                   7654 3210
                                      ----
n = 63 = 32 + 16 + 8 + 4 + 2 + 1 = 0011 1111
                                 & 1110 0011 => mask3 = 1110 0000 | 0000 0011
                                   ---------
Answer =                           0010 0011 => 32 + 2 + 1 = 35
mask1 = 1110 0000 = (~0) << 5 = (~0) << (j + 1)
mask2 = 0000 0011 = ~((~0) << 2) = ~((~0) << i)
      | ---------
        1110 0011 => mask3
```

```python
def clearRangeItoJ(n, i, j):
    mask1 = (~0) << (j + 1)
    mask2 = ~((~0) << i)
    mask3 = mask1 | mask2
    return n & mask3
print(clearRangeItoJ(63, 2, 4))
print(clearRangeItoJ(63, 1, 4))
```

```python
Counting the number of 1 bits of a given number n
if n = 20 = 16 + 4 = 0001 0100 => output is 2
if n = 55 = 32 + 16 + 4 + 2 + 1 = 0011 0111 => output is 5
```

```python
def countBits(n):
    ans = 0
    while(n):
        ans += (n & 1)    # 0001 0100 & 0000 0001
        n >>= 1   # n = n >> 1
        print ("Looping...")
    return ans
print(countBits(20))
print()
print(countBits(55))
```

```python
Counting the number of 1 bits of a given number n
if n = 20 = 16 + 4 = 0001 0100 => output is 2
if n = 22 = 16 + 4 + 2 = 0001 0110 =>  output is 3
if n = 55 = 32 + 16 + 4 + 2 + 1 = 0011 0111 => output is 5
ans   n
-------------------------------------
0     22 = 10110
1     22 & 21 = 10110 & 10101 = 10100 = 20
2     20 & 19 = 10100 & 10011 = 10000 = 16
3     16 & 15 = 10000 & 01111 = 00000 = 0
```

```python
def countBitsFast(n):
    ans = 0
    while(n):
        n = n & (n - 1)
        ans += 1
        print("Looping...")
    return ans
print(countBitsFast(20))
print()
print(countBitsFast(22))
print()
print(countBitsFast(55))
```

```python
def power_optimized(a, n):
    result = 1
    while (n > 0):
        last_bit = n & 1
        if (last_bit):
            result = result * a
        a = a * a
        n = n >> 1
        print ("Looping...")
    return result
print(power_optimized(2, 10))
```

```python
# LeetCode: 231. Power of Two
# Link: https://leetcode.com/problems/power-of-two/
class Solution(object):   # 30 ms
    def isPowerOfTwo(self, n):
        """
        :type n: int
        :rtype: bool
        """
        p = 0
        while (2 ** p < n):
            p = p + 1
            
        if (2 ** p == n):
            return True
        else:
            return False
```

```python
class Solution(object):    # 15 ms
    def isPowerOfTwo(self, n):
        """
        :type n: int
        :rtype: bool
        """
        # return n != 0 and n & (n - 1) == 0
        return n > 0 and not(n & (n - 1))
```
