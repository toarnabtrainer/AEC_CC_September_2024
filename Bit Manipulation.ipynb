# **Bit Manipulation**

```python
Bitwise Operators:
AND &, OR |, XOR ^, NOT ~, LEFT SHIFT >>, RIGHT SHIFT <<
```

```python
data1 = 100 => 64 + 32 + 4 = 0110 0100
data2 = 50 =>  32 + 16 + 2 = 0011 0010
data1 | data2 = 0110 0100
              | 0011 0010
                ---------
                0111 0110 => 64 + 32 + 16 + 4 + 2 = 118
```

```python
data1 = 100
data2 = 50
print (f"{data1} | {data2} = {data1 | data2}...")
```

```python
data1 = 100 => 64 + 32 + 4 = 0110 0100
data2 = 50 =>  32 + 16 + 2 = 0011 0010
data1 & data2 = 0110 0100
              & 0011 0010
                ---------
                0010 0000 = 32
```

```python
data1 = 100
data2 = 50
print (f"{data1} & {data2} = {data1 & data2}...")
```

```python
data1 = 100 => 64 + 32 + 4 = 0110 0100
data2 = 50 =>  32 + 16 + 2 = 0011 0010
data1 ^ data2 = 0110 0100
              ^ 0011 0010
                ---------
                0101 0110 => 64 + 16 + 4 + 2 = 86
```

```python
data1 = 100
data2 = 50
print (f"{data1} ^ {data2} = {data1 ^ data2}...")
```

```python
data1 = 100 => 64 + 32 + 4 = 0110 0100
~data1 = 1001 1011 = -101
         0110 0100 (1's complement)
                +1
        ----------
         0110 0101 (2's complement) => 64 + 32 + 4 + 1 = 101
```

```python
data1 = 100
print (f"~{data1} = {~data1}...")
```

```python
data1 = 100 => 64 + 32 + 4 = 0110 0100
data2 = data1 << 2
    0110 0100 << 2 =
  0110 010000 => 01 1001 0000 => 256 + 128 + 16 = 400
Conclusion: y << n = y * (2^n)
```

```python
data1 = 100
print (f"{data1} << 2 = {data1 << 2}...")
print (f"{data1} << 3 = {data1 << 3}...")
print (f"{data1} << 4 = {data1 << 4}...")
```

```python
data1 = 100 => 64 + 32 + 4 = 0110 0100
data2 = data1 >> 2
    0110 0100 >> 2 =
    000110 01 => 0001 1001 => 16 + 8 + 1 = 25
Conclusion: y >> n = y // (2^n)
```

```python
data1 = 100
print (f"{data1} >> 2 = {data1 >> 2}...")
data1 = 400
print (f"{data1} >> 3 = {data1 >> 3}...")
data1 = 1600
print (f"{data1} >> 4 = {data1 >> 4}...")
```

```python
20 = 16 + 4     = 0001 0100 (Even)
21 = 16 + 4 + 1 = 0001 0101 (Odd)
                & 0000 0001
                  ---------
                  0000 0001 (Non-Zero => ODD)
```

```python
def isOdd(n):
    return ((n & 1) == 1)
print(isOdd(20))
print(isOdd(21))
```

```python
def isEven(n):
    return not isOdd(n)
print(isEven(20))
print(isEven(21))
```

```python
Finding the i-th bit of the number n
                   7654 3210
n = 20 => 16 + 4 = 0001 0100
i = 2            & 0000 0100 => mask => 1 << 2 = 1 << i
                   ---------
                   0000 0100 != 0 => at the i-th place the bit is 1
                                7654 3210
n = 55 => 32 + 16 + 4 + 2 + 1 = 0011 0111
i = 5                         & 0010 0000 => mask => 1 << 5 = 1 << i
                                ---------
                                0010 0000 != 0 => at the i-th place the bit is 1
```

```python
def getBit(n, i):
    mask = 1 << i
    bit = 1 if (n & mask) else 0
    return bit
print(getBit(20, 2))
print(getBit(55, 5))
print(getBit(55, 3))
```
