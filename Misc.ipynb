```python
# LeetCode: 9. Palindrome Number
# Link: https://leetcode.com/problems/palindrome-number/
class Solution(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        if (x < 0): return False
        temp = 0
        nums = x
        while(nums):
            temp = temp * 10 + nums % 10
            nums //= 10
        return temp == x

print(Solution().isPalindrome(121))
print(Solution().isPalindrome(-121))
print(Solution().isPalindrome(10))
```

```python
# LeetCode: 69. Sqrt(x)
# Link: https://leetcode.com/problems/sqrtx/  # 1158 ms
class Solution(object):
    def mySqrt(self, x):
        """
        :type x: int
        :rtype: int
        """
        if (x == 0): return 0
        n = 1
        while(n * n < x):
            n += 1
        return n - 1 if n * n > x else n
print(Solution().mySqrt(4))
print(Solution().mySqrt(8))
print(Solution().mySqrt(20))
```

```python
# LeetCode: 69. Sqrt(x)
# Link: https://leetcode.com/problems/sqrtx/  # 22 ms
class Solution(object):
    def mySqrt(self, x):
        """
        :type x: int
        :rtype: int
        """
        low = 1
        high = x + 1
        ans = 0
        while (low < high):
            mid = (low + high) // 2
            if (mid * mid <= x):
                low = mid + 1
                ans = mid
            else:
                high = mid
        return ans
print(Solution().mySqrt(4))
print(Solution().mySqrt(8))
print(Solution().mySqrt(20))
```

```python
# LeetCode: 7. Reverse Integer
# Link: https://leetcode.com/problems/reverse-integer/   # 13 ms
class Solution(object):
    def reverse(self, x):
        """
        :type x: int
        :rtype: int
        """
        if (x >= -(2 ** 31) and x <= (2 ** 31 - 1)):
            x = str(x)
            if (x[0] == '-'):   # for -ve numbers
                a = int('-' + x[-1:0:-1])
            else:   # for +ve numbers
                a = int(x[::-1])
            if (a >= -(2 ** 31) and a <= (2 ** 31 - 1)):
                return a
            else:
                return 0
        else:
            return 0
print (Solution().reverse(123))
print (Solution().reverse(-123))
print (Solution().reverse(120))
```

```python
# Leetcode: 1. Two Sum
# Link: https://leetcode.com/problems/two-sum/  # O(n^2) 2178 ms
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        nums_size = len(nums)
        for i in range(0, nums_size - 1):
            for j in range(i + 1, nums_size):
                if (target == nums[i] + nums[j]):
                    return [i, j]
nums = [2,7,11,15]; target = 9
print(Solution().twoSum(nums, target))  # [0, 1]
nums = [3,2,4]; target = 6
print(Solution().twoSum(nums, target))  # [1, 2]
nums = [3,3]; target = 6
print(Solution().twoSum(nums, target))  # [0, 1]
```

```python
# Leetcode: 1. Two Sum
# Link: https://leetcode.com/problems/two-sum/  # O(n) 399 ms
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        for i in range(len(nums)):
            element = nums[i]
            data = target - element
            if (data in nums):
                position = nums.index(data)
                if (position != i):
                    return [i, position]

nums = [2,7,11,15]; target = 9
print(Solution().twoSum(nums, target))  # [0, 1]
nums = [3,2,4]; target = 6
print(Solution().twoSum(nums, target))  # [1, 2]
nums = [3,3]; target = 6
print(Solution().twoSum(nums, target))  # [0, 1]
```

```python
# LeetCode: 122. Best Time to Buy and Sell Stock II
# Link: Notepad.pw Link: https://notepad.pw/p5Achs3EAtvkDLsbUxOg  # 66 ms
class Solution(object):
    def maxProfit(self, prices):
        """
        :type prices: List[int]
        :rtype: int
        """
        profit = 0
        for i in range(1, len(prices)):
            if (prices[i] - prices[i - 1] > 0):
                profit += prices[i] - prices[i - 1]
        return profit
        
prices = [7,1,5,3,6,4]  # answer = 7
print(Solution().maxProfit(prices))
prices = [1,2,3,4,5]    # answer = 4
print(Solution().maxProfit(prices))
prices = [7,6,4,3,1]    # answer = 0
print(Solution().maxProfit(prices))
```
