```python
# LeetCode: 9. Palindrome Number
# Link: https://leetcode.com/problems/palindrome-number/
class Solution(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        if (x < 0): return False
        temp = 0
        nums = x
        while(nums):
            temp = temp * 10 + nums % 10
            nums //= 10
        return temp == x

print(Solution().isPalindrome(121))
print(Solution().isPalindrome(-121))
print(Solution().isPalindrome(10))
```

```python
# LeetCode: 69. Sqrt(x)
# Link: https://leetcode.com/problems/sqrtx/  # 1158 ms
class Solution(object):
    def mySqrt(self, x):
        """
        :type x: int
        :rtype: int
        """
        if (x == 0): return 0
        n = 1
        while(n * n < x):
            n += 1
        return n - 1 if n * n > x else n
print(Solution().mySqrt(4))
print(Solution().mySqrt(8))
print(Solution().mySqrt(20))
```

```python
# LeetCode: 69. Sqrt(x)
# Link: https://leetcode.com/problems/sqrtx/  # 22 ms
class Solution(object):
    def mySqrt(self, x):
        """
        :type x: int
        :rtype: int
        """
        low = 1
        high = x + 1
        ans = 0
        while (low < high):
            mid = (low + high) // 2
            if (mid * mid <= x):
                low = mid + 1
                ans = mid
            else:
                high = mid
        return ans
print(Solution().mySqrt(4))
print(Solution().mySqrt(8))
print(Solution().mySqrt(20))
```

```python
# LeetCode: 7. Reverse Integer
# Link: https://leetcode.com/problems/reverse-integer/   # 13 ms
class Solution(object):
    def reverse(self, x):
        """
        :type x: int
        :rtype: int
        """
        if (x >= -(2 ** 31) and x <= (2 ** 31 - 1)):
            x = str(x)
            if (x[0] == '-'):   # for -ve numbers
                a = int('-' + x[-1:0:-1])
            else:   # for +ve numbers
                a = int(x[::-1])
            if (a >= -(2 ** 31) and a <= (2 ** 31 - 1)):
                return a
            else:
                return 0
        else:
            return 0
print (Solution().reverse(123))
print (Solution().reverse(-123))
print (Solution().reverse(120))
```
