```python
Fill up one matrix with dimension nxn (where n is user given ODD integer)
with numbers ranging from 1 to n^2 using them once in order in the clockwise
manner starting from the left top corner of the matrix.

n = 5 (User given ODD integer)  row => row number, col => column number
(row) 0   1   2   3   4 (col)   rr => increment or decrement factor along the row (0, 1, -1)
-------------------------       cc => increment or decrement factor along the col (0, 1, -1)
 0 |  1   2   3   4   5         row = row + rr, col = col + cc
 1 | 16  17  18  19   6           rr  cc     tt = rr           1. Limit check
 2 | 15  24  25  20   7         ----------   rr = cc              if (col == n)
 3 | 14  23  22  21   8            0   1     cc = -tt             if (row == n)
 4 | 13  12  11  10   9            1   0                          if (col == -1)
-------------------------          0  -1     rr, cc = cc, -rr
                                  -1   0                       2. Pre-Occupied check
                                ----------                        if (arr[row][col] != 0)
                                   0   1
```

```python
while (True):
    n = int(input("Please enter ODD dimension integer number:"))
    if (n % 2 == 1): break
arr = [[0 for _ in range(n)] for _ in range(n)]
row = col = rr = 0; cc = 1
for data in range(1, n*n + 1):
    arr[row][col] = data
    tr = row + rr; tc = col + cc
    if (tc == n or tr == n or tc == -1 or arr[tr][tc] != 0):
        rr, cc = cc, -rr
    row += rr
    col += cc
for row in range(n):
    for col in range(n):
        print ("%4d"%(arr[row][col]), end = " ")
    print()
print ("End of the program...")
```

```python
Fill up one matrix with dimension nxn (where n is user given ODD integer)
with numbers ranging from 1 to n^2 using them once in order in the anti-clockwise
manner starting from the left top corner of the matrix.

n = 5 (User given ODD integer)  row => row number, col => column number
(row) 0   1   2   3   4 (col)   rr => increment or decrement factor along the row (0, 1, -1)
-------------------------       cc => increment or decrement factor along the col (0, 1, -1)
 0 |  1  16  15  14  13         row = row + rr, col = col + cc
 1 |  2  17  24  23  12           rr  cc     tt = rr           1. Limit check
 2 |  3  18  25  22  11         ----------   rr = -cc             if (row == n)
 3 |  4  19  20  21  10            1   0     cc = tt              if (col == n)
 4 |  5   6   7   8   9            0   1                          if (row == -1)
-------------------------         -1   0     rr, cc = -cc, rr
                                   0  -1                       2. Pre-Occupied check
                                ----------                        if (arr[row][col] != 0)
                                   1   0
```

```python
while (True):
    n = int(input("Please enter ODD dimension integer number:"))
    if (n % 2 == 1): break
arr = [[0 for _ in range(n)] for _ in range(n)]
row = col = cc = 0; rr = 1
for data in range(1, n*n + 1):
    arr[row][col] = data
    tr = row + rr; tc = col + cc
    if (tr == n or tc == n or tr == -1 or arr[tr][tc] != 0):
        rr, cc = -cc, rr
    row += rr
    col += cc
for row in range(n):
    for col in range(n):
        print ("%4d"%(arr[row][col]), end = " ")
    print()
print ("End of the program...")
```

```python
# LeetCode: 59. Spiral Matrix II
# Link: https://leetcode.com/problems/spiral-matrix-ii/
class Solution(object):
    def generateMatrix(self, n):
        """
        :type n: int
        :rtype: List[List[int]]
        """
        arr = [[0 for _ in range(n)] for _ in range(n)]
        row = col = rr = 0; cc = 1
        for data in range(1, n*n + 1):
            arr[row][col] = data
            tr = row + rr; tc = col + cc
            if (tc == n or tr == n or tc == -1 or arr[tr][tc] != 0):
                rr, cc = cc, -rr
            row += rr
            col += cc
        return arr
print (Solution().generateMatrix(5))
```
